```python
LAST_NAME = "Wales"
STUDENT_ID = "TUPM-26-7925"
FAVORITE_ARTIST = "IVOS"

id_digits = [int(c) for c in STUDENT_ID if c.isdigit()]
letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]

SEED_NUM = id_digits[-1]    # CHECK this against the SEED_NUM rule in your Lab 4 handout

print("SEED_NUM        :", SEED_NUM)
print("LAST_NAME       :", LAST_NAME)
print("FAVORITE_ARTIST :", FAVORITE_ARTIST)
```

```python
customers = [f"{LAST_NAME}-C{i}" for i in range(1, 6)]
prices = {"NOTEBOOK": 45, "PEN": 12, "USB": 350, "HEADPHONES": 899, "CALCULATOR": 650}
products = list(prices)

counter = 0
def next_quantity():
    """Quantity (1-5) from SEED_NUM and the next character position of FAVORITE_ARTIST."""
    global counter
    pos = counter % len(letters) + 1
    letter_value = ord(letters[pos - 1]) - ord("A") + 1
    counter += 1
    return (letter_value * pos + SEED_NUM) % 5 + 1

# each customer buys three different products
purchases = []
for i, customer in enumerate(customers):
    for j in range(3):
        purchases.append((customer, products[(i + j) % 5]))

# one repeated customer-product purchase (first customer buys the first product again)
purchases.append((customers[0], products[0]))

records = []
for n, (customer, product) in enumerate(purchases, 1):
    qty = next_quantity()
    records.append({"id": f"T{n:03d}", "customer": customer, "product": product,
                    "quantity": qty, "price": prices[product], "total": qty * prices[product]})

print("Fixed prices:", prices)
print()
print(f"{'ID':<6}{'Customer':<10}{'Product':<12}{'Qty':>4}{'Price':>7}{'Total':>8}")
for r in records:
    print(f"{r['id']:<6}{r['customer']:<10}{r['product']:<12}{r['quantity']:>4}{r['price']:>7}{r['total']:>8}")
```

```python
by_customer, by_product = {}, {}
for r in records:
    by_customer.setdefault(r["customer"], []).append(r)
    by_product.setdefault(r["product"], []).append(r)

print("By customer:")
for c, items in by_customer.items():
    print(f"  {c}: {[(i['product'], i['quantity']) for i in items]}")

print("\nBy product:")
for p, items in by_product.items():
    print(f"  {p}: {[(i['customer'], i['quantity']) for i in items]}")
```

```python
from collections import Counter

pair_counts = Counter((r["customer"], r["product"]) for r in records)
repeated = [(pair, c) for pair, c in pair_counts.items() if c > 1]

print("Total transactions        :", len(records))
print("Distinct customer-product pairs:", len(pair_counts))
print("Repeated customer-product purchases:")
for (customer, product), c in repeated:
    print(f"  {customer} bought {product} {c} times")
print()
print("Distinct customers:", sorted({r["customer"] for r in records}))
print("Distinct products :", sorted({r["product"] for r in records}))
```

```python
print(f"{'ID':<6}{'Customer':<10}{'Product':<12}{'Qty x Price':>14}{'Total':>8}")
for r in records:
    print(f"{r['id']:<6}{r['customer']:<10}{r['product']:<12}{str(r['quantity']) + ' x ' + str(r['price']):>14}{r['total']:>8}")

overall_total = sum(r["total"] for r in records)
print("\nOverall dataset value:", overall_total)
```

```python
def summarize(groups):
    out = {}
    for name, items in groups.items():
        total = sum(i["total"] for i in items)
        out[name] = {"transactions": len(items),
                     "units": sum(i["quantity"] for i in items),
                     "total": total,
                     "avg": total / len(items)}
    return out

customer_summary = summarize(by_customer)
product_summary = summarize(by_product)

print(f"{'Customer':<10}{'Trans':>6}{'Units':>7}{'Total':>8}{'Avg/Trans':>11}")
for k, s in customer_summary.items():
    print(f"{k:<10}{s['transactions']:>6}{s['units']:>7}{s['total']:>8}{s['avg']:>11.2f}")

print(f"\n{'Product':<12}{'Trans':>6}{'Units':>7}{'Revenue':>9}{'Avg/Trans':>11}")
for k, s in product_summary.items():
    print(f"{k:<12}{s['transactions']:>6}{s['units']:>7}{s['total']:>9}{s['avg']:>11.2f}")
```

```python
print("Customers ranked by total spent:")
for rank, (c, s) in enumerate(sorted(customer_summary.items(), key=lambda x: x[1]["total"], reverse=True), 1):
    print(f"  {rank}. {c:<10}{s['total']:>8}")

print("\nProducts ranked by revenue:")
for rank, (p, s) in enumerate(sorted(product_summary.items(), key=lambda x: x[1]["total"], reverse=True), 1):
    print(f"  {rank}. {p:<12}{s['total']:>8}")

print("\nTransactions ordered by value (highest first):")
for r in sorted(records, key=lambda r: r["total"], reverse=True):
    print(f"  {r['id']:<6}{r['customer']:<10}{r['product']:<12}{r['total']:>8}")
```

```python
top_customer = max(customer_summary, key=lambda c: customer_summary[c]["total"])
top_product = max(product_summary, key=lambda p: product_summary[p]["total"])
largest = max(records, key=lambda r: r["total"])

print("=== FINAL CONSOLIDATED TRANSACTION REPORT ===")
print("Total transactions        :", len(records))
print("Total units sold          :", sum(r["quantity"] for r in records))
print("Overall transaction value :", overall_total)
print("Average transaction value :", round(overall_total / len(records), 2))
print("Largest transaction       :", largest["id"], "-", largest["customer"], largest["product"], largest["total"])
print("Top customer              :", top_customer, "->", customer_summary[top_customer]["total"])
print("Top product               :", top_product, "->", product_summary[top_product]["total"])
print("Repeated purchases        :", [(pair, c) for pair, c in repeated])
```
